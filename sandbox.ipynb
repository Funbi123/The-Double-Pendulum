```python
#testing determinism chaos theroy for simple systems

import numpy as np
from matplotlib.animation import FuncAnimation
import matplotlib.pyplot as plt

m = 1
g = 9.8

# -------------------------
# Projectile 1
# -------------------------
x1, y1 = 0, 0
vx1, vy1 = 5, 10

t_of_flight1 = 2 * vy1 / g

x_1 = []
y_1 = []

for t in np.arange(0, t_of_flight1, 0.001):
    x_n = vx1 * t
    y_n = vy1 * t - 0.5 * g * t**2

    x_1.append(x_n)
    y_1.append(y_n)


# -------------------------
# Projectile 2
# -------------------------
x2, y2 = 0, 0
vx2, vy2 = 5, 10.01

t_of_flight2 = 2 * vy2 / g

x_2 = []
y_2 = []

for t in np.arange(0, t_of_flight2, 0.001):
    x_n = vx2 * t
    y_n = vy2 * t - 0.5 * g * t**2

    x_2.append(x_n)
    y_2.append(y_n)


# -------------------------
# Axis limits
# -------------------------

xmin = min(min(x_1), min(x_2)) - 0.5
xmax = max(max(x_1), max(x_2)) + 0.5

ymin = min(min(y_1), min(y_2)) - 0.5
ymax = max(max(y_1), max(y_2)) + 0.5


fig, ax = plt.subplots(figsize=(8, 8))

ax.set_xlim(xmin, xmax)
ax.set_ylim(ymin, ymax)
ax.set_aspect("equal")

# Two trajectories
traj1, = ax.plot(
    [],
    [],
    "-",
    color="tab:red",
    label="Projectile 1"
)

traj2, = ax.plot(
    [],
    [],
    "-",
    color="tab:blue",
    label="Projectile 2"
)

ax.legend()


# -------------------------
# Animation
# -------------------------

def update(frame):

    # Projectile 1
    if frame < len(x_1):
        traj1.set_data(
            x_1[:frame + 1],
            y_1[:frame + 1]
        )

    # Projectile 2
    if frame < len(x_2):
        traj2.set_data(
            x_2[:frame + 1],
            y_2[:frame + 1]
        )

    return traj1, traj2


# Need enough frames for the longer trajectory
frames = max(len(x_1), len(x_2))

animation = FuncAnimation(
    fig,
    update,
    frames=frames,
    interval=5,
    blit=True
)

plt.show()
```
